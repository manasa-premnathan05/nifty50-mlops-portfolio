# Experiment No. 7: CI/CD Pipeline with Open-Source Tools
### **Domain:** Financial Machine Learning Operations (MLOps)
### **Automation Framework:** GitHub Actions, Pytest, DVC, Docker
### **Artifacts Tracked:** Decision Tree Classifier (`best_model.pkl`), Dataset (`final_nifty50_dataset.xlsx`)

---

## **Aim & Objectives**
* **Aim:** Design and construct an automated Continuous Integration and Continuous Deployment (CI/CD) pipeline using open-source tools (GitHub Actions, Pytest, DVC) to enforce code quality, automate unit/integration testing, verify data/model versioning, and validate container builds upon code changes.
* **Key Objectives:**
  1. **CI/CD Pipeline Architecture:** Author a multi-stage GitHub Actions workflow (`.github/workflows/ci_cd.yml`) defining trigger conditions (`push`, `pull_request` to `main`).
  2. **Automated Code Linting & Static Analysis:** Integrate Flake8 and Black to catch syntax errors, enforce PEP 8 style standards, and detect undefined variables.
  3. **Automated Unit & API Test Suite:** Develop modular Pytest suites testing model inference contracts (feature dimensions, probability bounds) and FastAPI endpoints (`/health`, `/predict`, `/predict_batch`, negative tests).
  4. **Data & Model Version Control (DVC):** Configure DVC pipelines (`dvc.yaml`) to verify dependency tracking, model artifact integrity, and data provenance.
  5. **Container Build Verification:** Integrate container syntax checking and smoke test jobs in CI to prevent broken deployment images.

---


## 1. GitHub Actions Workflow Configuration

We inspect the complete declarative GitHub Actions workflow YAML file located at `.github/workflows/ci_cd.yml`.
The workflow features 4 orchestrated jobs:
1. `code-quality`: Flake8 syntax and lint inspection.
2. `unit-and-api-testing`: Pytest execution across model and API test suites.
3. `dvc-data-verification`: Validates artifact existence, feature consistency, and model integrity.
4. `docker-build`: Tests Docker container compilation and smoke test validation.

```python
# Inspect GitHub Actions Workflow Configuration
workflow_path = os.path.join(".github", "workflows", "ci_cd.yml")
assert os.path.exists(workflow_path), f"Workflow file not found at {workflow_path}!"

with open(workflow_path, "r", encoding="utf-8") as f:
    workflow_yaml = f.read()

print(f"GitHub Actions Workflow (.github/workflows/ci_cd.yml) - {len(workflow_yaml.splitlines())} lines:")
print("=" * 65)
print(workflow_yaml)
```

## 2. Automated Test Suite Architecture

Our test framework is organized under `tests/` and structured into two core testing suites:
1. `test_model_inference.py`: Unit tests validating model loading, 16-feature input contracts, output shape conformity, and probability calibrations.
2. `test_api_endpoints.py`: Integration tests exercising FastAPI endpoints (`/health`, `/predict`, `/predict_batch`, and HTTP 422 negative handling).

```python
# Display Test Suite Files
print("--- tests/test_model_inference.py ---")
with open(os.path.join("tests", "test_model_inference.py"), "r", encoding="utf-8") as f:
    print(f.read())

print("\n--- tests/test_api_endpoints.py ---")
with open(os.path.join("tests", "test_api_endpoints.py"), "r", encoding="utf-8") as f:
    print(f.read())
```

## 3. Local CI Execution & Test Suite Verification

We execute Pytest locally within the CI environment, verifying that all unit tests and API integration tests pass with zero errors.

```python
import subprocess

# Execute pytest on tests/ directory
res = subprocess.run([sys.executable, "-m", "pytest", "tests/", "-v", "--tb=short"], capture_output=True, text=True)
print("PYTEST EXECUTION LOGS:")
print("=" * 65)
print(res.stdout)
if res.stderr:
    print("STDERR:", res.stderr)

assert res.returncode == 0, "Pytest suite failed!"
print(">>> ALL 8 CI TEST CASES PASSED SUCCESSFULLY!")
```

## 4. Open-Source Data & Model Versioning with DVC

To decouple large binary datasets and model weights from git history while maintaining strict version provenance, we define our DVC stage manifest (`dvc.yaml`).
Here we inspect `dvc.yaml` and verify model/dataset tracking integrity.

```python
# Inspect DVC Configuration
with open("dvc.yaml", "r", encoding="utf-8") as f:
    dvc_content = f.read()

print("DVC Stage Manifest (dvc.yaml):")
print("=" * 65)
print(dvc_content)

# Verify DVC tracked dependencies exist
deps = ["final_nifty50_dataset.xlsx", "best_model.pkl"]
for d in deps:
    exists = os.path.exists(d)
    size = os.path.getsize(d) if exists else 0
    print(f"[DVC Dependency] {d:<30} -> Present: {exists} ({size:,} bytes)")
```
